# Session 4: Argument Parsing with `argparse`

A minimal demo of the stdlib `argparse` module. In a notebook there is no real command line, so we pass the arguments to `parse_args()` as a list, exactly as `sys.argv[1:]` would contain them.

## 1. Build a parser

```python
import argparse

parser = argparse.ArgumentParser(prog="greet", description="Greet someone from the command line.")
parser.add_argument("name", help="who to greet")                                        # positional argument
parser.add_argument("-n", "--count", type=int, default=1, help="how many times to greet")  # option with a value
parser.add_argument("-v", "--verbose", action="store_true", help="print extra output")    # flag (True/False)
```

## 2. Parse some arguments

```python
args = parser.parse_args(["Alice", "--count", "3", "-v"])
args
```

```python
if args.verbose:
    print(f"Greeting {args.name} {args.count} time(s)")
for _ in range(args.count):
    print(f"Hello, {args.name}!")
```

## 3. Defaults

```python
parser.parse_args(["Bob"])
```

## 4. The help text is generated for free

```python
parser.print_help()
```

## 5. Bad input

`argparse` validates types and exits with an error. Inside a notebook, that exit shows up as a `SystemExit` we can catch.

```python
try:
    parser.parse_args(["Alice", "--count", "three"])
except SystemExit as e:
    print("exit code:", e.code)
```
